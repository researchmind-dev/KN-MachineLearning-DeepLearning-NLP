```python
## Independent and dependent features
X=df[['Weight']] ### independent features should be data frame or 2 dimesnionalarray
y=df['Height'] ## this variiable can be in series or 1d array
```

```python
X_series=df['Weight']
np.array(X_series).shape
```

```python
np.array(y).shape
```

```python
## Train Test Split
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42) #everytime similar data will be picked when we used random_state
```

```python
## Standardization,  different columns have different unit(e.g kg , cm)
# so to make impact similar to we used standardization using zscore for each column
# where mean=0 and standard deviation =1
# This standardization applied for input features only as 
# we are going to apply gradient desent and not on output(dependent features)
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
```

```python
# so for test data we used transform , not fit_transform as 
# in train data ,from fit_transform, we get mean and SD and want to used same mean and SD from train data for test data
# so we just apply transform for test data we used same mean and SD from train data
# if we apply fit_transform to test data also , we get new mean and SD for test data also 
# which we don't want as it leads to data leakage problem
X_test=scaler.transform(X_test)
```

```python
X_test
```

```python
## Apply Simple Linear Regression
from sklearn.linear_model import LinearRegression
```

```python
# Whenever we need to used any function/API of sklearn , we need to first inilitiase a object 
# n_jobs = -1 means using all processors
regression=LinearRegression(n_jobs=-1)
```

```python
regression.fit(X_series,y_train)
```

```python
# here in .fir , regression need 2D array not 1D, so instead of X_train , if we used X_Series , we will get error
regression.fit(X_train,y_train)
```

```python
print("Coefficient or slope:",regression.coef_)
print("Intercept:",regression.intercept_)
```

```python
## plot Training data plot best fit line
plt.scatter(X_train,y_train)
plt.plot(X_train,regression.predict(X_train))
```

### prediction of test data
1. predicted height output= intercept +coef_(Weights)
2. y_pred_test =156.470 + 17.29(X_test)

```python
## Prediction for test data
y_pred=regression.predict(X_test)
```

```python
## Performance Metrics
from sklearn.metrics import mean_absolute_error,mean_squared_error
```

```python
mse=mean_squared_error(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
rmse=np.sqrt(mse)
print(mse)
print(mae)
print(rmse)
```

## R square 
Formula

**R^2 = 1 - SSR/SST**


R^2	=	coefficient of determination
SSR	=	sum of squares of residuals
SST	=	total sum of squares

```python
from sklearn.metrics import r2_score
```

```python
score=r2_score(y_test,y_pred)
print(score)
```

**Adjusted R2 = 1 – [(1-R2)*(n-1)/(n-k-1)]**

where:

R2: The R2 of the model
n: The number of observations
k: The number of predictor variables

```python
#display adjusted R-squared
1 - (1-score)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```
