### Named Entity Recognition

````python
sentence="The Eiffel Tower was built from 1887 to 1889 by Gustave Eiffel, whose company specialized in building metal frameworks and structures."
````

````python
import os
import nltk

# Define the local directory in your current project folder
local_nltk_dir = os.path.join(os.getcwd(), 'nltk_data')

# Download to the project folder
nltk.download('punkt', download_dir=local_nltk_dir)
nltk.download('punkt_tab', download_dir=local_nltk_dir)

# Tell NLTK to look for data in this local folder
if local_nltk_dir not in nltk.data.path:
    nltk.data.path.append(local_nltk_dir)
"""
In newer versions of NLTK (3.9+), 
sentence tokenization relies on language-specific tables found in the punkt_tab package 
rather than the legacy punkt model alone.
"""
````

````python
import nltk
words=nltk.word_tokenize(sentence)
print(words)
````

````python
tag_elements=nltk.pos_tag(words)
print(tag_elements)
````
