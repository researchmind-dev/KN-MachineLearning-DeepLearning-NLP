## Wordnet Lemmatizer
Lemmatization technique is like stemming. The output we will get after lemmatization is called ‘lemma’, which is a root word rather than root stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us understand it with an example −


```python
## Q&A,chatbots,text summarization
from nltk.stem import WordNetLemmatizer
```

```python
# WordNetLemmatizer requires the wordnet corpus (and optionally omw-1.4 for multilingual wordnet support) required in new version
import os
import nltk
# Set local download directory
local_nltk_dir = os.path.join(os.getcwd(), 'nltk_data')

# Download required corpora to the local folder
nltk.download('wordnet', download_dir=local_nltk_dir)
nltk.download('omw-1.4', download_dir=local_nltk_dir)

# Ensure NLTK looks into your local folder
if local_nltk_dir not in nltk.data.path:
    nltk.data.path.append(local_nltk_dir)
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
'''
POS- 
Noun-n (default, if not given pos)
verb-v
adjective-a
adverb-r
'''
print(lemmatizer.lemmatize("going",pos='n'))
print(lemmatizer.lemmatize("going",pos='v'))
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized", "Congrats"]
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='v'))
```
