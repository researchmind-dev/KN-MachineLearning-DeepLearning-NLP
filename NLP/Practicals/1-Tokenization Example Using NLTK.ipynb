```python
import os
import nltk

# Define the local directory in your current project folder
local_nltk_dir = os.path.join(os.getcwd(), 'nltk_data')

# Download to the project folder
nltk.download('punkt', download_dir=local_nltk_dir)
nltk.download('punkt_tab', download_dir=local_nltk_dir)

# Tell NLTK to look for data in this local folder
if local_nltk_dir not in nltk.data.path:
    nltk.data.path.append(local_nltk_dir)
"""
In newer versions of NLTK (3.9+), 
sentence tokenization relies on language-specific tables found in the punkt_tab package 
rather than the legacy punkt model alone.
"""
```

```python
corpus="""Hello Welcome,to Krish Naik's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""
```

```python
print(corpus)
```

```python
##  Tokenization
## 1st step : Paragraphs--to-->Sentences
from nltk.tokenize import sent_tokenize
```

```python
sent_tokenize(corpus,language='english') # if don't give language arg , its default to english
```

```python
documents=sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## 2nd step: Word Tokenization 
## Paragraph-->words
## sentence--->words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize

# with just word_tokenize 's is whole word but with wordpunct_tokenize ' and s are two word
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
# before this full stop (.) treated a diff word but with treebanktokenzier , full stop is part of its prev word itself 
# and only for the last full stop is treated as seperate word
```

```python
tokenizer=TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
