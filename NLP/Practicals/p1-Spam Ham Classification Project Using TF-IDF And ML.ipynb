```python
import pandas as pd
messages=pd.read_csv('SpamClassifier-master/smsspamcollection/SMSSpamCollection',
                    sep='\t',names=["label","message"])
```

```python
messages
```

```python
# import nltk
# nltk.download('stopwords')

import os
import nltk

# Define the local directory in your current project folder
local_nltk_dir = os.path.join(os.getcwd(), 'nltk_data')

# Download stopwords to the local folder
nltk.download('stopwords', download_dir=local_nltk_dir)

# Ensure NLTK searches this local directory
if local_nltk_dir not in nltk.data.path:
    nltk.data.path.append(local_nltk_dir)
```

```python
## Data Cleaning And Preprocessing
import re
# import nltk
# nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
corpus=[]
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-z]',' ',messages['message'][i])
    review=review.lower()
    review=review.split()
    review=[ps.stem(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
```

```python
corpus
```

## Test and Train Split

```python
## Output Features
y=pd.get_dummies(messages['label']) # label:  spam/ham is output
```

```python
y
```

```python
y=y.iloc[:,0].values # either ham or spam can be used as output feature ,so lets takes 0th ham as output
```

```python
y.shape
```

```python
## Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(corpus,y,test_size=0.20)
```

```python
len(X_train),len(y_train)
```

```python
X_train
```

### Create Bag Of Words

```python
## Create the Bag OF Words model
from sklearn.feature_extraction.text import CountVectorizer
## for Binary BOW enable binary=True
cv=CountVectorizer(max_features=2500,ngram_range=(1,2))
```

```python
## independent features
X_train=cv.fit_transform(X_train).toarray()
X_test=cv.transform(X_test).toarray()
```

```python
import numpy as np
np.set_printoptions(edgeitems=30, linewidth=100000, formatter=dict(float=lambda x: "%.3g" % x))
X_train
```

```python
cv.vocabulary_
```

### Train The model

```python
from sklearn.naive_bayes import MultinomialNB # Multinomial Naive Bayes
```

```python
spam_detect_model=MultinomialNB().fit(X_train,y_train)
```

```python
y_pred=spam_detect_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

### Creating The TF-IDF Model

```python
## Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(corpus,y,test_size=0.20)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tv=TfidfVectorizer(max_features=2500,ngram_range=(1,2))
X_train=tv.fit_transform(X_train).toarray()
X_test=tv.transform(X_test).toarray()
```

```python
X_train
```

```python
tv.vocabulary_
```

```python
from sklearn.naive_bayes import MultinomialNB
spam_tfidf_model = MultinomialNB().fit(X_train,y_train)
```

```python

#prediction
y_pred=spam_tfidf_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report
score = accuracy_score(y_test,y_pred)
print(score)
print(classification_report(y_test,y_pred))
```
